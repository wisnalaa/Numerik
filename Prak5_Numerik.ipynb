```python
def bisection(f, a, b, tol):
    if f(a) * f(b) > 0:
        print("Tidak ada akar di interval ini.")
        return None
    while (b - a) / 2 > tol:
        c = (a + b) / 2
        if f(c) == 0:
            return c
        elif f(a) * f(c) < 0:
            b = c
        else:
            a = c
    return (a + b) / 2
```

```python
def regula_falsi(f, a, b, tol):
    if f(a) * f(b) > 0:
        print("Tidak ada akar di interval ini.")
        return None
    xr = a
    while True:
        x_old = xr
        xr = a - (f(a) * (b - a)) / (f(b) - f(a))
        if abs(xr - x_old) < tol:
            break
        if f(a) * f(xr) < 0:
            b = xr
        else:
            a = xr
    return xr


f = lambda x: x**3 - x - 2
print("Akar Bisection:", bisection(f, 1, 2, 1e-6))
print("Akar Regula Falsi:", regula_falsi(f, 1, 2, 1e-6))
```

```python
def bisection(f, a, b, tol):
    if f(a) * f(b) > 0:
        print("Tidak ada akar di interval ini.")
        return None

    iterasi = 0
    print("Iterasi\t a\t\t b\t\t c\t\t f(c)")
    print("-" * 60)

    while (b - a) / 2 > tol:
        c = (a + b) / 2
        print(f"{iterasi}\t {a:.6f}\t {b:.6f}\t {c:.6f}\t {f(c):.6f}")

        if f(c) == 0:
            # f(c) tepat nol, berarti c adalah akar
            return c
        elif f(a) * f(c) < 0:
            b = c
        else:
            a = c

        iterasi += 1

    hasil = (a + b) / 2
    print(f"\nAkar hampiran: {hasil:.6f}")
    return hasil

def regula_falsi(f, a, b, tol):
    if f(a) * f(b) > 0:
        print("Tidak ada akar di interval ini.")
        return None

    iterasi = 0
    xr = a
    print("Iterasi\t a\t\t b\t\t xr\t\t f(xr)")
    print("-" * 60)

    while True:
        x_old = xr
        xr = (a * f(b) - b * f(a)) / (f(b) - f(a))
        print(f"{iterasi}\t {a:.6f}\t {b:.6f}\t {xr:.6f}\t {f(xr):.6f}")

        if abs(xr - x_old) < tol:
            break

        if f(a) * f(xr) < 0:
            b = xr
        else:
            a = xr

        iterasi += 1

    print(f"\nAkar hampiran: {xr:.6f}")
    return xr

f = lambda x: x**3 - x - 2

print("\n=== Metode Bisection ===")
akar_bisec = bisection(f, 1, 2, 1e-6)

print("\n=== Metode Regula Falsi ===")
akar_rf = regula_falsi(f, 1, 2, 1e-6)

print(f"\nAkar Bisection     = {akar_bisec:.6f}")
print(f"Akar Regula Falsi  = {akar_rf:.6f}")
```

```python
# Implementasi dan eksperimen untuk Tugas 3
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Fungsi metode Bisection dan Regula Falsi (dengan pencatatan iterasi dan nilai tiap langkah)
def bisection(f, a, b, tol, max_iter=10000):
    fa = f(a); fb = f(b)
    if fa * fb > 0:
        raise ValueError("Tidak ada perubahan tanda di interval.")
    iter_count = 0
    logs = []
    while (b - a) / 2.0 > tol and iter_count < max_iter:
        c = (a + b) / 2.0
        fc = f(c)
        logs.append((iter_count, a, b, c, fc))
        if fc == 0:
            return c, iter_count+1, logs
        if fa * fc < 0:
            b = c; fb = fc
        else:
            a = c; fa = fc
        iter_count += 1
    c = (a + b) / 2.0
    return c, iter_count, logs

def regula_falsi(f, a, b, tol, max_iter=10000):
    fa = f(a); fb = f(b)
    if fa * fb > 0:
        raise ValueError("Tidak ada perubahan tanda di interval.")
    iter_count = 0
    xr = a
    logs = []
    while True and iter_count < max_iter:
        x_old = xr
        xr = (a * fb - b * fa) / (fb - fa)
        fxr = f(xr)
        logs.append((iter_count, a, b, xr, fxr))
        if iter_count>0 and abs(xr - x_old) < tol:
            return xr, iter_count+1, logs
        # If first iter might also satisfy
        if abs(fxr) < 1e-18:
            return xr, iter_count+1, logs
        if fa * fxr < 0:
            b = xr; fb = fxr
        else:
            a = xr; fa = fxr
        iter_count += 1
        # safety stopping if interval small enough
        if abs(b-a)/2.0 < tol:
            return xr, iter_count, logs
    return xr, iter_count, logs

# Fungsi yang diberikan
f1 = lambda x: math.exp(-x) - x
f2 = lambda x: x**3 - 9*x + 3

# 1.a Analitik untuk f1: tunjukkan interval
# Kita lihat f1(0)=1, f1(1)=e^-1 -1 <0 => ada perubahan tanda pada [0,1].
interval_f1 = (0.0, 1.0)

# Temukan referensi akar presisi tinggi (menggunakan regula_falsi dengan tol sangat kecil)
root_f1_ref, _, _ = regula_falsi(f1, interval_f1[0], interval_f1[1], tol=1e-14, max_iter=10000)
# Untuk f2, pertama cari interval berubah tanda (scan)
xs = np.linspace(-10,10,20001)
sign_changes = []
for i in range(len(xs)-1):
    if f2(xs[i]) * f2(xs[i+1]) < 0:
        sign_changes.append((xs[i], xs[i+1]))
# compress contiguous tiny intervals
intervals_f2 = sign_changes

# Untuk acuan akar f2, kita cari akar pada tiap interval via regula falsi tol 1e-14
roots_f2_ref = []
used = set()
for (a,b) in intervals_f2:
    try:
        r,_,_ = regula_falsi(f2,a,b,tol=1e-14, max_iter=10000)
        # deduplicate by proximity
        if not any(abs(r - rr) < 1e-10 for rr in roots_f2_ref):
            roots_f2_ref.append(r)
    except Exception:
        pass
roots_f2_ref = sorted(roots_f2_ref)

# Percobaan toleransi
tols = [1e-1, 1e-3, 1e-5, 1e-7]
results = []

# Fungsi pembantu untuk menjalankan eksperimen dan mengukur galat relatif terhadap referensi
def run_experiments(f, intervals, root_refs, name):
    for (a,b), root_ref in zip(intervals, root_refs if isinstance(root_refs, list) else [root_refs]):
        for method in ['Bisection','Regula Falsi']:
            for tol in tols:
                try:
                    if method=='Bisection':
                        root, iters, logs = bisection(f, a, b, tol, max_iter=10000)
                    else:
                        root, iters, logs = regula_falsi(f, a, b, tol, max_iter=10000)
                    # hitung galat akhir sebagai |root - root_ref| dan |f(root)|
                    err_root = abs(root - root_ref)
                    err_f = abs(f(root))
                    results.append({
                        "Fungsi": name,
                        "Metode": method,
                        "Toleransi": tol,
                        "Jumlah Iterasi": iters,
                        "Nilai Akar": root,
                        "Galat_Root": err_root,
                        "Galat_f": err_f
                    })
                except Exception as e:
                    results.append({
                        "Fungsi": name,
                        "Metode": method,
                        "Toleransi": tol,
                        "Jumlah Iterasi": None,
                        "Nilai Akar": None,
                        "Galat_Root": None,
                        "Galat_f": None,
                        "Error": str(e)
                    })

# Jalankan untuk f1 (satu interval)
run_experiments(f1, [interval_f1], root_f1_ref, "e^(-x) - x")

# Jalankan untuk f2 (mungkin beberapa interval)
# compress intervals_f2 to unique intervals where sign changes detected
# We'll reduce intervals by grouping contiguous tiny spans: pick representative points from sign_changes every time sign changes occurs
intervals_f2_unique = []
i = 0
while i < len(intervals_f2):
    a = intervals_f2[i][0]
    b = intervals_f2[i][1]
    # skip contiguous ones
    j = i+1
    while j < len(intervals_f2) and abs(intervals_f2[j][0] - b) < 1e-6:
        b = intervals_f2[j][1]
        j += 1
    intervals_f2_unique.append((a,b))
    i = j

# compute reference roots for each unique interval
roots_f2_refs_unique = []
for (a,b) in intervals_f2_unique:
    try:
        r,_,_ = regula_falsi(f2,a,b,tol=1e-14, max_iter=10000)
        roots_f2_refs_unique.append(r)
    except Exception as e:
        roots_f2_refs_unique.append(None)

run_experiments(f2, intervals_f2_unique, roots_f2_refs_unique, "x^3 - 9x + 3")

# Tampilkan tabel menggunakan pandas
df = pd.DataFrame(results)
# Tampilkan dataframe rapi
display(df)

# Buat plot: toleransi vs jumlah iterasi (log scale for tol on x-axis)
plt.figure(figsize=(8,5))
# prepare plotting: for each function+method, collect tols and iters (only where iters not None)
groups = df.groupby(["Fungsi","Metode"])
for (func,met), group in groups:
    x = [math.log10(t) for t in group["Toleransi"]]
    y = [it if it is not None else np.nan for it in group["Jumlah Iterasi"]]
    plt.plot(x, y, marker='o', label=f"{func} - {met}")

plt.xlabel("log10(Toleransi)")
plt.ylabel("Jumlah Iterasi")
plt.title("Hubungan Toleransi dan Jumlah Iterasi")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# Simpulan awal otomatis:
summary = []
# For f1, find toleransi yang memenuhi <=100 iter and 4 decimal accuracy (|err_root| < 5e-5)
df_f1 = df[df["Fungsi"]=="e^(-x) - x"].copy()
for method in ["Bisection","Regula Falsi"]:
    sub = df_f1[df_f1["Metode"]==method]
    valid = sub[(sub["Jumlah Iterasi"]<=100) & (sub["Galat_Root"]<5e-5)]
    if not valid.empty:
        # choose largest tol (least strict) satisfying
        best = valid.sort_values("Toleransi", ascending=False).iloc[0]
        summary.append(f"{method} pada f1: toleransi optimal ≈ {best['Toleransi']} (iter={best['Jumlah Iterasi']}, galat_root={best['Galat_Root']:.2e})")
    else:
        summary.append(f"{method} pada f1: tidak ditemukan toleransi dari set percobaan yang memenuhi batasan.")

# For f2, check each root separately (group by root reference rounding)
df_f2 = df[df["Fungsi"]=="x^3 - 9x + 3"].copy()
# group by Metode and root_ref by matching nearest root in intervals_f2_unique roots list
for method in ["Bisection","Regula Falsi"]:
    sub = df_f2[df_f2["Metode"]==method]
    # For each unique root (approx), check same criteria
    for rref in roots_f2_refs_unique:
        if rref is None:
            continue
        subr = sub.copy()
        # select rows whose root is close to rref
        subr = subr[subr["Nilai Akar"].notnull()]
        subr = subr[abs(subr["Nilai Akar"] - rref) < 1e-3]
        valid = subr[(subr["Jumlah Iterasi"]<=100) & (subr["Galat_Root"]<5e-5)]
        if not valid.empty:
            best = valid.sort_values("Toleransi", ascending=False).iloc[0]
            summary.append(f"{method} pada f2, akar≈{rref:.6f}: toleransi optimal ≈ {best['Toleransi']} (iter={best['Jumlah Iterasi']}, galat_root={best['Galat_Root']:.2e})")
        else:
            summary.append(f"{method} pada f2, akar≈{rref:.6f}: tidak ditemukan toleransi dari set percobaan yang memenuhi batasan.")

# Print brief summary
for s in summary:
    print(s)
```
