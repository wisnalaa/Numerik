```python
import numpy as np
def f(x):
  return x**3 - 4*x + 1
```

```python
def trapezoidal(a, b, h):
  n = int((b - a) / h)
  x = np.linspace(a, b, n+1)
  y = f(x)
  return h/2 * (y[0] + 2*np.sum(y[1:-1]) + y[-1])
```

```python
def simpson_13(a, b, h):
  n = int((b-a)/h)
  if n % 2 != 0:
    raise ValueError("n harus genap")
  h = (b - a) / n
  x = np.linspace(a, b, n+1)
  y = f(x)
  return h/3 * (y[0] + 4*np.sum(y[1:-1:2]) + 2*np.sum(y[2:-2:2]) + y[-1])
```

```python
def simpson_38(a, b, h):
  n = int((b-a)/h)
  if n % 3 != 0:
    raise ValueError("n harus kelipatan 3")
  h = (b - a) / n
  x = np.linspace(a, b, n+1)
  y = f(x)
  return 3*h/8 * (y[0] + 3*np.sum(y[1:-1][(np.arange(1,n)%3)!=0]) + 2*np.sum(y[3:-3:3]) + y[-1])
```

```python
# Parameter
a, b = 0, 4
I_exact = 36
h_values = [1, 0.5, 0.25]

print("Metode\t\t h\t Hasil\t\t Error Absolut")
print("-"*55)

for h in h_values:
    trap = trapezoidal(a, b, h)
    print(f"Trapesium\t {h}\t {trap:.6f}\t {abs(I_exact - trap):.6f}")

    if int((b-a)/h) % 2 == 0:
        s13 = simpson_13(a, b, h)
        print(f"Simpson 1/3\t {h}\t {s13:.6f}\t {abs(I_exact - s13):.6f}")

    if int((b-a)/h) % 3 == 0:
        s38 = simpson_38(a, b, h)
        print(f"Simpson 3/8\t {h}\t {s38:.6f}\t {abs(I_exact - s38):.6f}")
```
