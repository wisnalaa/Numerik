```python
import numpy as np

# Matriks yang akan difaktorisasi LU
A = np.array([[4, 3, 2],
              [2, 1, 3],
              [3, 3, 4]])

# Ukuran matriks
n = len(A)

# Inisialisasi matriks L dan U
L = np.zeros((n, n))
U = np.zeros((n, n))

# Faktorrisasi LU
for i in range(n):
    # Mengisi elemen diagonal utama matriks L dengan 1
    L[i, i] = 1

    # Menghitung elemen-elemen matriks U
    for j in range(i, n):
        U[i, j] = A[i, j] - sum(L[i, k] * U[k, j] for k in range(i))

    # Menghitung elemen-elemen matriks L
    for j in range(i + 1, n):
        L[j, i] = (A[j, i] - sum(L[j, k] * U[k, i] for k in range(i))) / U[i, i]

# Hasil faktorisasi
print("Matriks L: ")
print(L)
print("Matriks U: ")
print(U)

# Verifikasi hasil faktorisasi
print("Hasil perkalian matriks L dan U: ")
print(np.dot(L, U))
```

```python
import numpy as np
from scipy.linalg import lu_factor, lu, lu_solve

# Matriks yang akan difaktorrisasi LU
A = np.array([[4, 3, 2],
              [2, 1, 3],
              [3, 3, 4]])

# vektor b
b = np.array([1, 2, 3])

# Faktorrisasi LU dan faktor faktorisasi
lu, piv = lu_factor(A)

# Solusi y dari Ly = b
y = lu_solve((lu, piv), b)

# Vektor Y
Y = np.array([4, 5, 6])

# Solusi x dari Ux = Y
x = lu_solve((lu, piv), Y)

# Hasil solusi
print("Solusi y: ")
print(y)
print("Solusi x: ")
print(x)
```

```python
import numpy as np
from scipy.linalg import lu_factor, lu_solve, lu

# Matriks A dan b
A21 = np.array([[2, 1, 1],
              [4, 3, 3],
              [8, 7, 9]])

b21 = np.array([1, 2, 3])

# 🔹 Dekomposisi LU secara eksplisit (untuk melihat L dan U)
P, L, U = lu(A21)

print("Matriks L:")
print(L)
print("\nMatriks U:")
print(U)
print("\nPerkalian P * L * U:")
print(P @ L @ U)  # cek hasilnya apakah kembali ke A11
print("\nApakah sama dengan A21?")
print(np.allclose(P @ L @ U, A21))  # True jika hasil sama (dalam batas toleransi float)

# 🔹 Dekomposisi untuk penyelesaian sistem persamaan
lu_factorized = lu_factor(A21)
y21 = lu_solve(lu_factorized, b21)

print("\nSolusi [x, y, z]:")
print(y21)
```
