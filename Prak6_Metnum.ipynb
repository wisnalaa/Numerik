```python
# Definisikan fungsi dan turunannya
def f(x):
    return x**2 - 2  # Contoh fungsi f(x) = x^2 - 2

def df(x):
    return 2*x  # Turunan pertama f'(x) = 2x

# Metode Newton-Raphson
def newton_raphson(x0, tol=1e-6, max_iter=100):
    x = x0
    for i in range(max_iter):
        fx = f(x)
        dfx = df(x)
        if dfx == 0:
            print("Turunan pertama f'(x) = 0, iterasi dihentikan.")
            return None
        x_new = x - fx / dfx
        if abs(x_new - x) < tol:
            return x_new
        x = x_new
    print("Tidak konvergen dalam jumlah iterasi maksimum.")
    return None

# Nilai awal dan toleransi
x0 = 1.0  # Nilai awal
akar = newton_raphson(x0)
print(f"Akar persamaan adalah: {akar}")
```

```python
import math

# Definisikan fungsi dan turunannya
def f(x):
    return math.sin(x) - 0.5  # f(x) = sin(x) - 0.5

def df(x):
    return math.cos(x)        # f'(x) = cos(x)

# Metode Newton-Raphson
def newton_raphson(x0, tol=1e-6, max_iter=100):
    x = x0
    for i in range(max_iter):
        fx = f(x)
        dfx = df(x)
        if dfx == 0:
            print("Turunan pertama f'(x) = 0, iterasi dihentikan.")
            return None
        x_new = x - fx / dfx
        if abs(x_new - x) < tol:
            print(f"Konvergen pada iterasi ke-{i+1}")
            return x_new
        x = x_new
    print("Tidak konvergen dalam jumlah iterasi maksimum.")
    return None

# Nilai awal dan toleransi
for x0 in [0.5, 2.0, 5.0]:  # Nilai awal
  print(f"Nilai awal: {x0}")
  akar = newton_raphson(x0)
  if akar is not None:
    print(f"Akar persamaan adalah: {akar}\n")
```
