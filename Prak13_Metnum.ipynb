```python
import numpy as np
import matplotlib.pyplot as plt
# Definisi Fungsi
def f(x):
  return np.exp(-x) * np.sin(2*x) + x**2
  # Definisi Turunan Analitik (Eksak) untuk verifikasi error
  # Turunan dari f(x) adalah f&#39;(x) = -e^(-x)sin(2x) + 2e^(-x)cos(2x) + 2x
  def f_exact(x):
    return -np.exp(-x) * np.sin(2*x) + 2*np.exp(-x) * np.cos(2*x) + 2*x
```

```python
def forward_diff(f, x, h):
  return (f(x + h) - f(x)) / h
def backward_diff(f, x, h):
  return (f(x) - f(x - h)) / h
def central_diff(f, x, h):
  return (f(x + h) - f(x - h)) / (2 * h)
```

```python
def f_exact(x):
  return -np.exp(-x) * np.sin(2*x) + 2*np.exp(-x) * np.cos(2*x) + 2*x

# Parameter
x_point = 2.0
h = 0.1

# Hitung Nilai
val_exact = f_exact(x_point)
val_fwd = forward_diff(f, x_point, h)
val_bwd = backward_diff(f, x_point, h)
val_ctr = central_diff(f, x_point, h)

# Hitung Error Relatif (%)
err_fwd = abs((val_fwd - val_exact) / val_exact) * 100
err_bwd = abs((val_bwd - val_exact) / val_exact) * 100
err_ctr = abs((val_ctr - val_exact) / val_exact) * 100

# Tampilkan Hasil
print(f"--- Hasil Perhitungan di x = {x_point} dengan h = {h} ---")
print(f"Nilai Eksak (Analitik) : {val_exact:.6f}")
print("-" * 50)
print(f"Metode Maju (Forward) : {val_fwd:.6f} | Error: {err_fwd:.4f}%")
print(f"Metode Mundur (Backward): {val_bwd:.6f} | Error: {err_bwd:.4f}%")
print(f"Metode Pusat (Central) : {val_ctr:.6f} | Error: {err_ctr:.4f}%")
```

```python
x_point = 2.0
h_values = [0.1, 0.01, 0.001]

val_exact = f_exact(x_point)

print("h\t\tForward Error (%)\tBackward Error (%)\tCentral Error (%)")
print("-"*75)

for h in h_values:
    val_fwd = forward_diff(f, x_point, h)
    val_bwd = backward_diff(f, x_point, h)
    val_ctr = central_diff(f, x_point, h)

    err_fwd = abs((val_fwd - val_exact)/val_exact)*100
    err_bwd = abs((val_bwd - val_exact)/val_exact)*100
    err_ctr = abs((val_ctr - val_exact)/val_exact)*100

    print(f"{h:.3f}\t\t{err_fwd:.6f}\t\t{err_bwd:.6f}\t\t{err_ctr:.6f}")
```

```python
x0 = 2
h = 0.01
slope = central_diff(f, x0, h)
y0 = f(x0)

x = np.linspace(1, 3, 100)
y = f(x)

# Persamaan garis singgung
tangent = slope * (x - x0) + y0

plt.plot(x, y, label="f(x)")
plt.plot(x, tangent, '--', label="Garis Singgung")
plt.scatter(x0, y0, color='red')

plt.legend()
plt.xlabel("x")
plt.ylabel("y")
plt.title("Kurva f(x) dan Garis Singgung di x = 2")
plt.grid()
plt.show()
```

```python
def f2(x):
    return np.exp(-x**2)

def f2_exact(x):
    return -2*x*np.exp(-x**2)

x = 1
h_vals = [1e-1, 1e-2, 1e-3]

print("h\t\tForward\t\tBackward\tCentral")
print("-"*60)

for h in h_vals:
    exact = f2_exact(x)
    ef = abs((forward_diff(f2, x, h)-exact)/exact)
    eb = abs((backward_diff(f2, x, h)-exact)/exact)
    ec = abs((central_diff(f2, x, h)-exact)/exact)
    print(f"{h:.0e}\t{ef:.6e}\t{eb:.6e}\t{ec:.6e}")
```
