```python
import numpy as np
import matplotlib.pyplot as plt
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 3, 5, 4, 6])
# PERHITUNGAN REGRESI LINIER (Least Square)

n = len(x)
sum_x = np.sum(x)
sum_y = np.sum(y)
sum_xy = np.sum(x * y)
sum_x2 = np.sum(x * x)

# rumus slope (b) dan intercept (a)
b = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x**2)
a = (sum_y - b * sum_x) / n

print("Persamaan Regresi:")
print(f"y = {a:.4f} + {b:.4f}x")
# MEMBUAT GARIS REGRESI
x_line = np.linspace(min(x), max(x), 100)
y_line = a + b * x_line
# GRAFIK
plt.figure(figsize=(8,5))

# scatter data asli
plt.scatter(x, y, color='blue', label='Data Asli')

# garis regresi
plt.plot(x_line, y_line, label='Garis Regresi', linewidth=2)

# tampilan grafik
plt.title('Regresi Linier - Metode Kuadrat Terkecil')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.legend()
plt.show()
```

#Latihan

1.	Data tinggi badan (x dalam cm) dan berat badan (y dalam kg):
   x: 150, 155, 160, 165, 170
   y: 45, 50, 55, 60, 65
Tentukan model regresi, grafik, dan maknanya.
2.	Data hubungan antara suhu ruangan (x) dengan konsumsi listrik AC (y):
   x: 20, 22, 24, 26, 28
   y: 300, 350, 400, 450, 500
   Hitung persamaan regresi linier dan grafik.


```python
import numpy as np
import matplotlib.pyplot as plt

def regresi_linier(x, y, title, xlabel, ylabel):
    # Konversi ke numpy array
    x = np.array(x)
    y = np.array(y)

    # Perhitungan regresi linier (Least Square)
    n = len(x)
    sum_x = np.sum(x)
    sum_y = np.sum(y)
    sum_xy = np.sum(x * y)
    sum_x2 = np.sum(x * x)

    # slope (b) dan intercept (a)
    b = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x**2)
    a = (sum_y - b * sum_x) / n

    # Tampilkan persamaan regresi
    print(f"Persamaan Regresi:")
    print(f"y = {a:.2f} + {b:.2f}x\n")

    # Garis regresi
    x_line = np.linspace(min(x), max(x), 100)
    y_line = a + b * x_line

    # Plot grafik
    plt.figure(figsize=(8,5))
    plt.scatter(x, y, label='Data Asli')
    plt.plot(x_line, y_line, label='Garis Regresi', linewidth=2)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.grid(True)
    plt.legend()
    plt.show()

    return a, b
```

```python
x1 = [150, 155, 160, 165, 170]
y1 = [45, 50, 55, 60, 65]

regresi_linier(
    x1, y1,
    title='Regresi Linier Tinggi vs Berat Badan',
    xlabel='Tinggi Badan (cm)',
    ylabel='Berat Badan (kg)'
)
```

```python
x2 = [20, 22, 24, 26, 28]
y2 = [300, 350, 400, 450, 500]

regresi_linier(
    x2, y2,
    title='Regresi Linier Suhu vs Konsumsi Listrik AC',
    xlabel='Suhu Ruangan (°C)',
    ylabel='Konsumsi Listrik (Watt)'
)
```
